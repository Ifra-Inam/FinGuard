```python
BASE_PATH = "/Volumes/finguard/landing/raw_files"
INITIAL_PATH = f"{BASE_PATH}/initial"

TRANSACTIONS_PATH = f"{INITIAL_PATH}/transactions/"
CUSTOMERS_PATH = f"{INITIAL_PATH}/customers/"
FRAUD_INVESTIGATIONS_PATH = f"{INITIAL_PATH}/fraud_investigations/"
```

```python
from pyspark.sql.functions import current_timestamp, col, lit

def ingest_to_bronze(source_path, target_table, batch_id):

    df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(source_path)

    df = df.withColumn("_ingestion_timestamp", current_timestamp()).withColumn("_source_file", col("_metadata.file_path")).withColumn("_batch_id", lit(batch_id))

    df.write.format("delta").mode("overwrite").saveAsTable(target_table)

    print(f"Successfully ingested {target_table}")
    print(f"Records: {df.count()}")
```

```python
ingest_to_bronze(TRANSACTIONS_PATH , "finguard.bronze.transactions", "initial")
ingest_to_bronze(CUSTOMERS_PATH , "finguard.bronze.customers", "initial")
ingest_to_bronze(FRAUD_INVESTIGATIONS_PATH, "finguard.bronze.fraud_investigation", "initial")
```
